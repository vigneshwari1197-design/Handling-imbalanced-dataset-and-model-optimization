# Import neccessary library

```python
from sklearn import datasets
import pandas as pd
```

# Data collection

```python
breast_cancer_data = datasets.load_breast_cancer()
```

```python
breast_cancer_data
```

```python
breast_cancer_data.feature_names
```

```python
print(breast_cancer_data.DESCR)
```

```python
breast_cancer_data_df = pd.DataFrame(data=breast_cancer_data.data,columns = breast_cancer_data.feature_names)
breast_cancer_data_df["Target"]=breast_cancer_data.target
breast_cancer_data_df
```

# Data understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

# Datapreparation not need for data cleaning

# model building

```python
X = breast_cancer_data_df.drop(labels =["Target"], axis=1)
X
```

```python
y = breast_cancer_data_df["Target"]
y
```

```python
X.shape,y.shape
```

# original data set proportional

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
plt.pie(x=y.value_counts(),autopct='%1.0f%%',labels = ["Benign cancer","Malignant cancer"])
plt.show()
```

```python
#Note--Stratify sampling
#Stratify=None means its differ b/w train and test not equal proportional value
#stratify=y its equally proportional value in original chart and after split also its same value
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=1234,shuffle=True,stratify=y )
```

```python
print("Traning sample:", X_train.shape,y_train.shape)
print("Testing sample:" ,X_test.shape,y_test.shape)
```

# After split we check the piechart
1.train and test

```python
from matplotlib import pyplot as plt
plt.pie(x=y_train.value_counts(),autopct='%1.0f%%',labels = ["Benign cancer","Malignant cancer"])
plt.show()
```

```python
from matplotlib import pyplot as plt
plt.pie(x=y_test.value_counts(),autopct='%1.0f%%',labels = ["Benign cancer","Malignant cancer"])
plt.show()
```

# model training

```python
%%time
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight={0: 2, 1: 1} )
dt_model.fit(X_train,y_train)
```

# model testing/model evaulation

## training data

```python
y_predict_train = dt_model.predict(X_train)
y_predict_train
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_train,y_predict_train))
```

```python
print(confusion_matrix(y_train,y_predict_train))
```

## test data

```python
y_predict_test = dt_model.predict(X_test)
y_predict_test
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_test,y_predict_test))
```

```python
print(confusion_matrix(y_test,y_predict_test))
```
